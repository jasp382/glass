## Join Table with one or more tables

```python
shp = '/mnt/x/phd_ca/ine_conversion/bd_centro_15jan25.xlsx'

shp_pk = 'code'

left_sheet = 'bd_freg'

right = {
    '/mnt/x/phd_ca/ine_conversion/pop91_freg22_v2.xlsx' : {
        'fk' : 'dicofre',
        'cols' : ['popRes91', 'pop0_14_91', 'pop65m_91', 'ie91', 'pp65m_91'],
        'sheet' : 'data'
    },
    '/mnt/x/phd_ca/ine_conversion/escola91_freg22_v2.xlsx' : {
        'fk' : 'dicofre',
        'cols' : [
            'nt15_91', '1ceb15_91', '2ceb15_91', '3ceb15_91',
            'sec15_91', 'sup15_91', 'emed_91', 'p15sup91'
        ], 
        'sheet' : 'data'
    }
}


out = '/mnt/x/phd_ca/ine_conversion/bd_centro_22jan25.xlsx'

from glass.tbl.joins import join_fields

join_fields(
    shp, shp_pk, right, out, leftSheet=left_sheet,
    _how='left', norelval=-1
)
```
