```python
rdv = '/home/jasp/escim/rdv_sec.shp'

facs = {
    200 : '/home/jasp/escim/regvec_fac_0_200.shp',
    500 : '/home/jasp/escim/regvec_fac_200_500.shp',
    2000 : '/home/jasp/escim/regvec_fac_m500.shp'
}

out = '/home/jasp/escim/'

bname = 'rdvsec'

import os
from glass.mob.grsete import con_facilities_to_net

i = 0
for f in facs:
    of = os.path.join(out, f'{bname}_{str(f)}.shp')
    if not i:
        rdv2 = con_facilities_to_net(
            rdv, facs[f], 'KPH', 
            'ONEWAY', "F_ELEV", "T_ELEV",
            of, threshold=f
        )

        i += 1

        continue

    rdv2 = con_facilities_to_net(
        rdv2, facs[f], 'kph', 'oneway',
        'f_elev', 't_elev',
        of, threshold=f
    )
```
