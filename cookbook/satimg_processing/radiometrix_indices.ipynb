```python
from glass.rst.sat.idx import CalcIndexes
from glass.pys.oss import lst_ff
from glass.cons.sat import get_lwibands
```

```python
ifolder = '/mnt/x/sattougi/imgs_v2/' 

idxs = {
    'evi' : '/mnt/x/sattougi/idxs_v2/evi3_20210708.tif'
}
```

```python
ibands = get_lwibands()

fbands = lst_ff(ifolder, file_format='.tif')

bands = {}
for i in ibands:
    for f in fbands:
        if i in f:
            bands[i] = f
            break

print(bands)
```

```python
calcidx = CalcIndexes('sentinel-2', 'grass')
```

```python
status = calcidx.read_data(bands)
```

```python
for idx in idxs:
    status = calcidx.calc_idx(idx)

    calcidx.export_result(idx, idxs[idx])
```
