```python
dem = '/home/jasp/csig/dem_3763.tif'

twirst = '/home/jasp/csig/twi_pt.tif'
```

```python
import os

from glass.pys.tm import now_as_str
from glass.pys.oss import fprop
from glass.wenv.grs import grass_session
from glass.it.rst   import rst_to_grs, grs_to_rst
from glass.rst.surf.wtr import twi
```

```python
ws, loc = os.path.dirname(twirst), now_as_str(utc=True)

# Create GRASS GIS session
gb = grass_session(ws, loc=loc, srs=dem)
```

```python
gdem =rst_to_grs(dem)

gtwi = twi(gdem, fprop(twirst, 'fn'))

grs_to_rst(gtwi, twirst, as_cmd=True, dtype='Float32', nodata=-1)
```
