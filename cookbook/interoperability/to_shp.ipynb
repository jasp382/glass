```python
from osgeo import ogr
gml_file = '/mnt/x/aosm_tk1p01/nl_pdok/LandCoverUnit.gml'
ds = ogr.Open(gml_file)
for i in range(ds.GetLayerCount()):
    layer = ds.GetLayerByIndex(i)
    geom_type = layer.GetGeomType()
    if geom_type == ogr.wkbNone:
        print(f"Layer '{layer.GetName()}' -> TABELA sem geometria")
    else:
        print(f"Layer '{layer.GetName()}' -> Geometria: {ogr.GeometryTypeToName(geom_type)}")
```
