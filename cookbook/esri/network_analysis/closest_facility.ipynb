```python
nd  = r'X:\rms\rms_pt.gdb\redeviaria\ndset_driving'

folder = r'X:\rms\acc_hsp21'

impedance = 'DriveTime'

incident_name = 'incidents.shp'
facility_name = 'facilities.shp'

oname = 'hsp_results'
```

```python
import os
from glass.pys.oss import lst_fld
from glass.esri.mob.na import closest_facility
```

```python
# Folders
folders = lst_fld(folder)

# Run closest facility
for fld in folders:
    closest_facility(
        nd, 
        os.path.join(fld, facility_name),
        os.path.join(fld, incident_name),
        os.path.join(fld, f'{oname}.dbf'),
        impedance_attr=impedance,
        oneway_name="Oneway"
        #oneway_name=None
    )
```
