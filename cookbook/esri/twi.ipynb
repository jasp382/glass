```python
dem = 'X:atrisk_fire1/topografia/concse/se_elev.tif'

twi = 'X:atrisk_fire1/topografia/concse/se_twi.tif'
```

```python
import os
from glass.pys.oss import mkdir
from glass.esri.rst.surf import slope
from glass.esri.rst.hydro import flow_direction, flow_accumulation
from glass.esri.rst.alg import rstcalc
from glass.esri.rd.rst import rst_to_lyr
from glass.esri.prop.rst import rst_geoprop
```

```python
ws = mkdir(os.path.dirname(twi), timerand=True, overwrite=None)
```

```python
lwleft, csize = rst_geoprop(dem)
```

```python
slope_path, slope_lyr = slope(
    dem, os.path.join(ws, 'slope.tif'),
    data='DEGREE'
)

dir_path, dir_lyr = flow_direction(dem, os.path.join(ws, 'direction.tif'))

acc_path, acc_lyr = flow_accumulation(dir_lyr, os.path.join(ws, 'accrst.tif'))
```

```python
tanslope_path, tanslope_lyr = rstcalc(
    [slope_lyr], ['x'],
    'Con(x > 0, Tan((x * 1.570796) / 90), 0.001)',
    os.path.join(ws, 'tan_slope.tif'),
    template = dem
)
```

```python
_, twi_lyr = rstcalc(
    [tanslope_lyr, acc_lyr], ['x', 'a'],
    f'Ln(((a + 1) * {str(csize)}) / x)',
    twi,
    template=dem
)
```
